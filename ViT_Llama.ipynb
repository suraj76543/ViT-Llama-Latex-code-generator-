```python
import os
from tqdm import tqdm
from PIL import Image as PILImage

from datasets import Dataset, Features, Image as HFImage, Value

dataset_folder = "/content/IMG2Latex/PRINTED_TEX_230k"
image_root = os.path.join(dataset_folder, "generated_png_images")

# img_names and formulas come from your text files
images_txt   = os.path.join(dataset_folder, "corresponding_png_images.txt")
formulas_txt = os.path.join(dataset_folder, "final_png_formulas.txt")

with open(images_txt, "r") as f:
    img_names = [line.strip() for line in f]

with open(formulas_txt, "r") as f:
    formulas = [line.strip() for line in f]

assert len(img_names) == len(formulas)

data_images = []
data_formulas = []

for img_name, formula in tqdm(zip(img_names, formulas), total=len(img_names)):
    # Normalize file name
    filename = os.path.basename(img_name)          # e.g. "05c58c888b106b0.png"
    img_path = os.path.join(image_root, filename)  # full path

    # Skip if file missing
    if not os.path.exists(img_path):
        continue

    # Check if image is readable; skip corrupted ones
    try:
        with PILImage.open(img_path) as im:
            im.verify()    # just verifies header, doesn't load full image
    except Exception:
        # corrupted / unreadable; skip it
        continue

    data_images.append(img_path)
    data_formulas.append(formula)

print("Valid samples:", len(data_images))

# Build HuggingFace Dataset from paths
features = Features({
    "image": HFImage(),              # will load from file paths
    "formula": Value("string"),
})

dataset1 = Dataset.from_dict(
    {"image": data_images, "formula": data_formulas},
    features=features,
)

print(dataset1)
print(dataset1[0]["image"], "\n", dataset1[0]["formula"])
```

```python
dataset = dataset1.select(range(2000))  # or more later

instruction = "Write the LaTeX representation for this image."

def preprocess_function(sample):
    conversation = [
        {
            "role": "user",
            "content": [
                {"type": "text", "text": instruction},
                {"type": "image", "image": sample["image"]},
            ],
        },
        {
            "role": "assistant",
            "content": [
                {"type": "text", "text": sample["formula"]},
            ],
        },
    ]

    # Apply chat template and tokenize
    chat_text = tokenizer.apply_chat_template(
        conversation,
        tokenize=False,
        add_generation_prompt=False,
    )

    # Tokenize text and image together
    tokenized_output = tokenizer(
        text=chat_text, # Explicitly pass chat_text as keyword arg 'text'
        images=[sample["image"]], # Explicitly pass images as keyword arg 'images'
        add_special_tokens=True,
        return_tensors="pt",
        padding="max_length", # Ensure padding if needed for map
        max_length=2048, # Use a max_length here
        truncation=True,
    )

    # Create labels - ignore padding
    labels = tokenized_output["input_ids"].clone().squeeze(0) # Remove batch dimension
    labels[labels == tokenizer.pad_token_id] = -100

    return {
        "input_ids": tokenized_output["input_ids"].squeeze(0).tolist(), # Remove batch dimension and convert to list
        "attention_mask": tokenized_output["attention_mask"].squeeze(0).tolist(), # Convert to list
        "pixel_values": tokenized_output["pixel_values"].squeeze(0), # Keep as tensor, remove batch dimension
        "labels": labels.tolist(), # Convert to list
    }

# Map the dataset to the new format, performing tokenization and image processing
dataset = dataset.map(
    preprocess_function,
    remove_columns=dataset.column_names, # Remove original columns, including 'messages'
    batched=False, # Process one by one
)

print(dataset[0])
```

```python
from typing import List, Dict, Any
from transformers import PreTrainedTokenizerBase
import torch

# The DataCollatorForVision will now be a simple padder, as tokenization is done in the dataset map
class DataCollatorForVision(object):
    def __init__(self, tokenizer: PreTrainedTokenizerBase):
        self.tokenizer = tokenizer

    def __call__(self, features: List[Dict[str, Any]]) -> Dict[str, Any]:
        # Features are already tokenized and have pixel_values
        input_ids = [feature["input_ids"] for feature in features]
        attention_mask = [feature["attention_mask"] for feature in features]
        labels = [feature["labels"] for feature in features]
        pixel_values = [feature["pixel_values"] for feature in features]

        # Pad input_ids and labels
        padded_input_ids = torch.nn.utils.rnn.pad_sequence(
            [torch.tensor(x) for x in input_ids], batch_first=True, padding_value=self.tokenizer.pad_token_id
        )
        padded_attention_mask = torch.nn.utils.rnn.pad_sequence(
            [torch.tensor(x) for x in attention_mask], batch_first=True, padding_value=0
        )
        padded_labels = torch.nn.utils.rnn.pad_sequence(
            [torch.tensor(x) for x in labels], batch_first=True, padding_value=-100
        )

        batch = {
            "input_ids": padded_input_ids,
            "attention_mask": padded_attention_mask,
            "labels": padded_labels,
            "pixel_values": torch.stack(pixel_values)
        }
        return batch
```

```python
FastVisionModel.for_training(model)  # very important

training_args = SFTConfig(
    per_device_train_batch_size = 1,
    gradient_accumulation_steps = 4,
    warmup_steps = 5,
    max_steps = 50,                  # start small, increase later
    learning_rate = 2e-4,
    fp16 = not is_bf16_supported(),
    bf16 = is_bf16_supported(),
    logging_steps = 1,
    optim = "adamw_8bit",
    weight_decay = 0.01,
    lr_scheduler_type = "linear",
    seed = 3407,
    output_dir = "outputs",
    report_to = "none",

    #  Required for vision fine-tuning like in Unsloth’s own examples:
    remove_unused_columns = False,
    dataset_text_field = None, # No longer needed as data is pre-tokenized
    dataset_kwargs = {"skip_prepare_dataset": True},
    dataset_num_proc = 4,
    max_seq_length = 2048, # Still relevant for internal checks
)

# Instantiate the custom data collator - now a simple padder
data_collator = DataCollatorForVision(tokenizer)

trainer = SFTTrainer(
    model = model,
    tokenizer = tokenizer,
    train_dataset = dataset,
    data_collator = data_collator,      # Use the instantiated data collator
    args = training_args,
)

trainer.train()
```
